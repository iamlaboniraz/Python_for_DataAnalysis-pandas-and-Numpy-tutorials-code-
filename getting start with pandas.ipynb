## What is a Series in Pandas?
- In Python, a Pandas Series is a one-dimensional (1D) labeled array that stores data.

- সহজ বাংলায়, Pandas Series হলো এমন একটি Data Structure যেখানে প্রতিটি Data বা Value-এর সাথে একটি Index থাকে।

- সাধারণত একটি DataFrame-এর একটি Column হলো একটি Pandas Series।

```python
import pandas as pd
```

```python
df = pd.Series([100,200,300,400])
df
```

```python
df.array
```

```python
df.index
```

```python
df1 = pd.Series([100,200,300,400], index=['Shirt','Pant','Pant','Shirt'])
```

```python
df1
```

```python
df1.index
```

# DataFrame 

```python
import pandas as pd

data = {
    'Product': ['Shirt', 'Jeans', 'T-shirt'],
    'Price': [500, 1200, 300],
    'Sales': [20, 15, 30]
}

df = pd.DataFrame(data)

df
```

```python
df['Sales']      # Series
df[['Sales']]    # DataFrame
```
